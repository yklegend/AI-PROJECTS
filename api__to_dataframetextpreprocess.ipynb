```python
genre_api_url = "https://api.themoviedb.org/3/genre/movie/list?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US"
genre_response = requests.get(genre_api_url)
genre_data = genre_response.json()['genres']

genre_map = {genre['id']: genre['name'] for genre in genre_data}

print("Genre mapping created:")
print(genre_map)
```

Now, let's regenerate the `df` DataFrame to include the `genre_ids` column from the API response.

```python
all_dfs = []
for i in range(1,429):
  response = requests.get(f"https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page={i}")
  # Include 'genre_ids' in the selection
  temp_df = pd.DataFrame(response.json()['results'])[['id','title','overview', 'genre_ids']]
  all_dfs.append(temp_df)

df = pd.concat(all_dfs, ignore_index=True)
display(df.head())
```

Now that we have the `genre_ids` in our `df`, let's convert them to genre names using the `genre_map` we created.

```python
def map_genre_ids_to_names(genre_ids_list, genre_map):
    if isinstance(genre_ids_list, list):
        return [genre_map.get(gid, 'Unknown') for gid in genre_ids_list]
    return []

df['genres'] = df['genre_ids'].apply(lambda x: map_genre_ids_to_names(x, genre_map))

# Drop the original 'genre_ids' column if no longer needed
df = df.drop(columns=['genre_ids'])

display(df.head())
```
